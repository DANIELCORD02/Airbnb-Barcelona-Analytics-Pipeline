```python
#importación de la librería pandas con el aleas pd
import pandas as pd
```

```python
#lectura del documento csv y asignandolo como df
df = pd.read_csv("listings.csv")
```

```python
# tomamos solamente las columnas que nos serán de utilidad
columnas_importantes = [
    'id', 'name','host_id','host_name',
    'neighbourhood_group_cleansed',
    'neighbourhood_cleansed',
    'latitude', 'longitude',
    'room_type', 'accommodates',
    'price', 'number_of_reviews','review_scores_rating'
]

df_limpio = df[columnas_importantes].copy()
df_limpio.head()
```

```python
# limpieza de data 
df_limpio.info()
```

```python
df_limpio['price'][100]
```

```python
#limpieza de la columna precio, quitamos el símbolo "$" y el separador ',' también lo convertimos a float
df_limpio['price'] = df_limpio['price'].replace({r'\$': '', ',': ''}, regex=True).astype(float)
```

```python
#llenamos los campos vacíos en la puntuación con 0, para facilitar la subida a SQL
df_limpio['review_scores_rating'] = df_limpio['review_scores_rating'].fillna(0) 
```

```python
# Diccionario con los nuevos nombres en español (formato SQL friendly)
nuevos_nombres = {
    'id': 'id_anuncio',
    'name': 'titulo_anuncio',
    'host_id': 'id_anfitrion',
    'host_name': 'nombre_anfitrion',
    'neighbourhood_group_cleansed': 'distrito',
    'neighbourhood_cleansed': 'barrio',
    'latitude': 'latitud',
    'longitude': 'longitud',
    'room_type': 'tipo_habitacion',
    'accommodates': 'capacidad_huespedes',
    'price': 'precio',
    'number_of_reviews': 'cantidad_resenas',
    'review_scores_rating': 'puntuacion'
}

# Aplicar el cambio al dataframe limpio
df_limpio = df_limpio.rename(columns=nuevos_nombres)

# Verificamos cómo quedó
print(df_limpio.head(2))
```

```python
print(df_limpio.head())
df_limpio.to_csv('listings_limpio.csv', index=False, encoding='cp1252', errors='replace')
print("Archivo limpio y guardado como 'listings_limpio.csv'")
```
